# Tipos e Acesso aos Elementos de um Array

```python
# criando um array
import numpy as np

arr = np.array([1,2,3,4,5])
print(f"elementos:{arr},Tipo dos elementos:{arr.dtype} , tamanho do array:{arr.shape}")

##Criando uma matrix 
matrix = np.arange(1.,10).reshape((3,3))
print(matrix)
print (f"tipos dos elementos:{matrix.dtype},Dimensoes da matriz:{matrix.shape}")

## Acessando a 2 linha :
linha = matrix[1]
print(linha)

## Acessando a 1 coluna:
coluna = matrix[:,1]
print(coluna)

```

# Multiplicacao utilizando apenas listas em Python

```python
import numpy as np

A = np.array([[1, 2], [3,4], [5,6]])
B = np.array([[1], [2]])
print(A)
print(B)


#Multiplicação das Matrizes

D = A @ B
print(D)
```

```python
import numpy as np

N = 500

A = np.arange(N*N , dtype = "int64").reshape(N,N)
B = A * 2

C = A @ B

print(C)
```

# Pandas:Series e DataFrames 


# Series

```python
import pandas as pd
cidades = [
    "Sao paulo (SP)",
    "Rio de Janeiro(RJ)",
    "Brasilia(DF)",
    "Fortaleza(CE)",
    "Salvador(BA)",
    "Belo horizonte",
    "Manaus(AM)",
    "Curitiba(PR)", 
]

populacoes = [
    12,
    6.5,
    3,
    2.4,
    2.4,
    2.3,
    2,
    1.7
   
]

cidades_series = pd.Series(cidades)
print(cidades_series)

populacoes_series = pd.Series(populacoes)
print(populacoes_series)

print(cidades_series[3])
print(cidades_series[3:5])
print(populacoes_series.max())#numero mais alto
print(populacoes_series.mean())#media

```

# DataFrames

```python
import pandas as pd

cidades_populosas = pd.DataFrame({"cidade": cidades_series,"populacao em milhoes":populacoes_series})
cidades_populosas
```
