# Exemplos de Operadores & Logicas 
## Aritméticos:


```python
##SOMA
#Dano
dano_espada = 10
dano_magico = 15                                                                       

total_de_dano = dano_espada + dano_magico 
print(total_de_dano)

##SUBTRAÇÃO
#Vida

vida_atual = 20
vida_maxima = 33

vida_perdida = vida_atual - vida_maxima #subtracao
print(vida_perdida)

##MULTIPLICAÇÃO
#Dano critico

dano_critico = total_de_dano * 2
print(dano_critico)

##DIVIZAO 
#XP do grupo

valor_totalXP = 1000
membros_grupo = 3
xp_por_membro = int(valor_totalXP / membros_grupo)
print(xp_por_membro)

```

## Operadores Relacionais (COMPARAÇÃO)

```python
teste_percepcao = 16 # resultado do "dado" do jogador
dificuldade_cd = 12  # CD exigida pelo jogo

print(teste_percepcao >= dificuldade_cd) #Conferi se o valor tirado no teste de percepcao foi MAIOR OU IGUAL  a dificuldade

classe_personagem = "Necromante "
print(classe_personagem == "feiticeiro") #Conferi se classe de personagem é IGUAL a feiticeiro
#False
print(classe_personagem != "Mago") #Conferi se a classe e DIFERENTE de Mago
#True
```

## Operadore Logicos (AND,OR,NOT)

```python
tem_chave = True
porta_destrancada = False

# not -> inverte o valor
esta_invisivel = False
if not esta_invisivel:
    print("Os inimigos podem te ver!")

# and -> as duas precisam ser verdadeiras
if tem_chave and not porta_destrancada:
    print("Você destranca a porta com a chave.")

# or -> basta uma ser verdadeira
carisma_alto = False
forca_alta = True

if carisma_alto or forca_alta:
    print("Você consegue convencer ou intimidar o goblin.")

```
